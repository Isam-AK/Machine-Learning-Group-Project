```python
# ── Imports ──────────────────────────────────────────────────────────────────
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image

import torch
import torch.nn as nn
import torchvision.models as models
import torchvision.transforms as transforms
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler
from sklearn.model_selection import train_test_split
from sklearn.utils.class_weight import compute_class_weight

print("All imports successful")
print(f"PyTorch version: {torch.__version__}")
```

```python
# -- Paths ---------------------------------------------------------------------
# archive/ lives in the parent of this notebook folder
NB_DIR   = os.getcwd()
DATA_DIR = os.path.abspath(os.path.join(NB_DIR, "..", "archive"))
if not os.path.isdir(DATA_DIR):
    DATA_DIR = os.path.abspath(os.path.join(NB_DIR, "archive"))

IMG_DIR_1     = os.path.join(DATA_DIR, "HAM10000_images_part_1")
IMG_DIR_2     = os.path.join(DATA_DIR, "HAM10000_images_part_2")
METADATA_PATH = os.path.join(DATA_DIR, "HAM10000_metadata.csv")

for p in [DATA_DIR, IMG_DIR_1, IMG_DIR_2, METADATA_PATH]:
    ok = os.path.exists(p)
    print(("OK " if ok else "MISSING ") + p)
```

```python
# ── Load Metadata ─────────────────────────────────────────────────────────────
df = pd.read_csv(METADATA_PATH)

print(f"Total images: {len(df)}")
print(f"\nColumns: {list(df.columns)}")
print(f"\nClass distribution:")
print(df['dx'].value_counts())
```

```python
# ── Label Mapping ─────────────────────────────────────────────────────────────
# Benign: nv, bkl, df, vasc
# Malignant/pre-malignant: mel, bcc, akiec

benign_classes    = ['nv', 'bkl', 'df', 'vasc']
malignant_classes = ['mel', 'bcc', 'akiec']

df['binary_label'] = df['dx'].apply(lambda x: 0 if x in benign_classes else 1)

print("Binary label distribution:")
print(df['binary_label'].value_counts())
print(f"\n0 = Benign, 1 = Malignant")
print(f"\nBenign total:    {(df['binary_label'] == 0).sum()}")
print(f"Malignant total: {(df['binary_label'] == 1).sum()}")
```

```python
# ── Build Image Paths ─────────────────────────────────────────────────────────
def find_image_path(image_id):
    for img_dir in [IMG_DIR_1, IMG_DIR_2]:
        path = os.path.join(img_dir, image_id + ".jpg")
        if os.path.exists(path):
            return path
    return None

df['image_path'] = df['image_id'].apply(find_image_path)

missing = df['image_path'].isna().sum()
print(f"Total images found: {len(df) - missing}")
print(f"Missing images:     {missing}")
```

```python
# ── Train / Validation / Test Split ───────────────────────────────────────────
# 70% train, 15% validation, 15% test
# Stratified to preserve class balance in each split

train_df, temp_df = train_test_split(
    df,
    test_size=0.30,
    random_state=42,
    stratify=df['binary_label']
)

val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    random_state=42,
    stratify=temp_df['binary_label']
)

print(f"Train size:      {len(train_df)} images")
print(f"Validation size: {len(val_df)} images")
print(f"Test size:       {len(test_df)} images")
print(f"\nTrain malignant %:      {100 * train_df['binary_label'].mean():.1f}%")
print(f"Validation malignant %: {100 * val_df['binary_label'].mean():.1f}%")
print(f"Test malignant %:       {100 * test_df['binary_label'].mean():.1f}%")

# Save split indices so Haruna and Caden use identical data
train_df.index.to_series().to_csv('train_indices.csv', index=False)
val_df.index.to_series().to_csv('val_indices.csv', index=False)
test_df.index.to_series().to_csv('test_indices.csv', index=False)

print("\n✓ Split indices saved to CSV")
```

```python
# ── Transforms ────────────────────────────────────────────────────────────────
# ImageNet mean and std — used because MobileNetV2 was pretrained on ImageNet
IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD  = [0.229, 0.224, 0.225]

train_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomVerticalFlip(),
    transforms.RandomRotation(20),
    transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2),
    transforms.ToTensor(),
    transforms.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD)
])

val_test_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD)
])

print("✓ Train transform:    resize + augmentation + normalise")
print("✓ Val/Test transform: resize + normalise only")
```

```python
# ── Custom Dataset ─────────────────────────────────────────────────────────────
class SkinLesionDataset(Dataset):
    def __init__(self, dataframe, transform=None, label_col='binary_label'):
        self.df        = dataframe.reset_index(drop=True)
        self.transform = transform
        self.label_col = label_col

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row   = self.df.iloc[idx]
        image = Image.open(row['image_path']).convert('RGB')
        label = int(row[self.label_col])

        if self.transform:
            image = self.transform(image)

        return image, label

# Create datasets
train_dataset = SkinLesionDataset(train_df, transform=train_transform)
val_dataset   = SkinLesionDataset(val_df,   transform=val_test_transform)
test_dataset  = SkinLesionDataset(test_df,  transform=val_test_transform)

print(f"✓ Train dataset:      {len(train_dataset)} images")
print(f"✓ Validation dataset: {len(val_dataset)} images")
print(f"✓ Test dataset:       {len(test_dataset)} images")

# Quick sanity check — load one image and check its shape
image, label = train_dataset[0]
print(f"\nSample image tensor shape: {image.shape}")
print(f"Sample label: {label} ({'Malignant' if label == 1 else 'Benign'})")
```

```python
# ── Class Weights & DataLoaders ───────────────────────────────────────────────
# Compute class weights to counter the 80/20 imbalance
class_weights = compute_class_weight(
    class_weight='balanced',
    classes=np.array([0, 1]),
    y=train_df['binary_label'].values
)
class_weights = torch.FloatTensor(class_weights)
print(f"Class weights — Benign: {class_weights[0]:.3f}, Malignant: {class_weights[1]:.3f}")

# WeightedRandomSampler — oversamples malignant during training
sample_weights = train_df['binary_label'].apply(
    lambda x: class_weights[x].item()
).values
sampler = WeightedRandomSampler(
    weights=sample_weights,
    num_samples=len(sample_weights),
    replacement=True
)

# DataLoaders
train_loader = DataLoader(train_dataset, batch_size=32, sampler=sampler)
val_loader   = DataLoader(val_dataset,   batch_size=32, shuffle=False)
test_loader  = DataLoader(test_dataset,  batch_size=32, shuffle=False)

print(f"\n✓ Train loader:      {len(train_loader)} batches")
print(f"✓ Validation loader: {len(val_loader)} batches")
print(f"✓ Test loader:       {len(test_loader)} batches")
```

```python
# ── MobileNetV2 Stage 1 — Binary Classifier ───────────────────────────────────
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# Load pretrained MobileNetV2
model_stage1 = models.mobilenet_v2(weights=models.MobileNet_V2_Weights.DEFAULT)

# Freeze all layers
for param in model_stage1.parameters():
    param.requires_grad = False

# Replace classification head for 2-class output
model_stage1.classifier = nn.Sequential(
    nn.Dropout(p=0.2),
    nn.Linear(1280, 2)
)

# Move model to device
model_stage1 = model_stage1.to(device)

# Count trainable parameters
trainable = sum(p.numel() for p in model_stage1.parameters() if p.requires_grad)
total     = sum(p.numel() for p in model_stage1.parameters())
print(f"\n✓ Model loaded and head replaced")
print(f"  Trainable parameters: {trainable:,}")
print(f"  Total parameters:     {total:,}")
print(f"  Frozen parameters:    {total - trainable:,}")
```

```python
# ── Loss Function & Optimiser (Warm-up Phase) ─────────────────────────────────
criterion = nn.CrossEntropyLoss(weight=class_weights.to(device))

# Only train the classifier head during warm-up
optimizer_warmup = torch.optim.Adam(
    model_stage1.classifier.parameters(),
    lr=1e-4
)

print("✓ Loss function: CrossEntropyLoss with class weights")
print("✓ Optimiser: Adam (head only, lr=1e-4)")
```

```python
# ── Training Loop ─────────────────────────────────────────────────────────────
from sklearn.metrics import recall_score
import time

def train_one_epoch(model, loader, optimizer, criterion, device):
    model.train()
    running_loss = 0.0
    all_preds, all_labels = [], []

    for images, labels in loader:
        images, labels = images.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = model(images)
        loss    = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item()
        preds = outputs.argmax(dim=1).cpu().numpy()
        all_preds.extend(preds)
        all_labels.extend(labels.cpu().numpy())

    epoch_loss   = running_loss / len(loader)
    epoch_recall = recall_score(all_labels, all_preds, pos_label=1)
    return epoch_loss, epoch_recall


def evaluate(model, loader, criterion, device):
    model.eval()
    running_loss = 0.0
    all_preds, all_labels = [], []

    with torch.no_grad():
        for images, labels in loader:
            images, labels = images.to(device), labels.to(device)
            outputs = model(images)
            loss    = criterion(outputs, labels)

            running_loss += loss.item()
            preds = outputs.argmax(dim=1).cpu().numpy()
            all_preds.extend(preds)
            all_labels.extend(labels.cpu().numpy())

    epoch_loss   = running_loss / len(loader)
    epoch_recall = recall_score(all_labels, all_preds, pos_label=1)
    return epoch_loss, epoch_recall


print("✓ Training and evaluation functions defined")
```

```python
# ── Warm-up Training (Head Only, 5 Epochs) ────────────────────────────────────
WARMUP_EPOCHS = 5
best_recall   = 0.0
history       = {'train_loss': [], 'val_loss': [], 'train_recall': [], 'val_recall': []}

print("Starting warm-up training (head only)...")
print(f"{'Epoch':<8} {'Train Loss':<12} {'Train Recall':<15} {'Val Loss':<12} {'Val Recall':<12}")
print("-" * 60)

for epoch in range(1, WARMUP_EPOCHS + 1):
    start = time.time()

    train_loss, train_recall = train_one_epoch(model_stage1, train_loader, optimizer_warmup, criterion, device)
    val_loss,   val_recall   = evaluate(model_stage1, val_loader, criterion, device)

    history['train_loss'].append(train_loss)
    history['val_loss'].append(val_loss)
    history['train_recall'].append(train_recall)
    history['val_recall'].append(val_recall)

    # Save best model based on validation recall
    if val_recall > best_recall:
        best_recall = val_recall
        torch.save(model_stage1.state_dict(), 'best_model_stage1.pt')
        saved = "✓ saved"
    else:
        saved = ""

    elapsed = time.time() - start
    print(f"{epoch:<8} {train_loss:<12.4f} {train_recall:<15.4f} {val_loss:<12.4f} {val_recall:<12.4f} {saved}")

print(f"\nWarm-up complete. Best validation recall: {best_recall:.4f}")
```

```python
# ── Fine-tuning (Unfreeze Last 4 Backbone Blocks) ─────────────────────────────
for param in model_stage1.features[15:].parameters():
    param.requires_grad = True

trainable = sum(p.numel() for p in model_stage1.parameters() if p.requires_grad)
print(f"✓ Unfrozen blocks 15-18")
print(f"  Trainable parameters now: {trainable:,}")

# New optimiser with different LRs for backbone vs head
optimizer_finetune = torch.optim.Adam([
    {'params': model_stage1.features[15:].parameters(), 'lr': 1e-5},
    {'params': model_stage1.classifier.parameters(),    'lr': 1e-4}
])

FINETUNE_EPOCHS = 5
print(f"\nStarting fine-tune training ({FINETUNE_EPOCHS} epochs)...")
print(f"{'Epoch':<8} {'Train Loss':<12} {'Train Recall':<15} {'Val Loss':<12} {'Val Recall':<12}")
print("-" * 60)

for epoch in range(1, FINETUNE_EPOCHS + 1):
    start = time.time()

    train_loss, train_recall = train_one_epoch(model_stage1, train_loader, optimizer_finetune, criterion, device)
    val_loss,   val_recall   = evaluate(model_stage1, val_loader, criterion, device)

    history['train_loss'].append(train_loss)
    history['val_loss'].append(val_loss)
    history['train_recall'].append(train_recall)
    history['val_recall'].append(val_recall)

    if val_recall > best_recall:
        best_recall = val_recall
        torch.save(model_stage1.state_dict(), 'best_model_stage1.pt')
        saved = "✓ saved"
    else:
        saved = ""

    elapsed = time.time() - start
    print(f"{epoch:<8} {train_loss:<12.4f} {train_recall:<15.4f} {val_loss:<12.4f} {val_recall:<12.4f} {saved}")

print(f"\nFine-tuning complete. Best validation recall: {best_recall:.4f}")
```

```python
# ── Plot Training Curves ───────────────────────────────────────────────────────
epochs = range(1, len(history['train_loss']) + 1)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Loss
ax1.plot(epochs, history['train_loss'], label='Train Loss')
ax1.plot(epochs, history['val_loss'],   label='Val Loss')
ax1.axvline(x=5.5, color='gray', linestyle='--', label='Warm-up / Fine-tune')
ax1.set_title('Stage 1 — Loss')
ax1.set_xlabel('Epoch')
ax1.set_ylabel('Loss')
ax1.legend()

# Recall
ax2.plot(epochs, history['train_recall'], label='Train Recall')
ax2.plot(epochs, history['val_recall'],   label='Val Recall')
ax2.axvline(x=5.5, color='gray', linestyle='--', label='Warm-up / Fine-tune')
ax2.axhline(y=0.95, color='red', linestyle=':', label='Target (0.95)')
ax2.set_title('Stage 1 — Recall')
ax2.set_xlabel('Epoch')
ax2.set_ylabel('Recall')
ax2.legend()

plt.tight_layout()
plt.savefig('stage1_training_curves.png', dpi=150)
plt.show()
print("✓ Training curves saved")
```

```python
    # ── Load Best Model & Evaluate on Test Set ────────────────────────────────────
from sklearn.metrics import (classification_report, confusion_matrix,
                             roc_auc_score, roc_curve)

# Load best checkpoint
model_stage1.load_state_dict(torch.load('best_model_stage1.pt', map_location=device))
model_stage1.eval()

all_preds, all_labels, all_probs = [], [], []

with torch.no_grad():
    for images, labels in test_loader:
        images = images.to(device)
        outputs = model_stage1(images)
        probs   = torch.softmax(outputs, dim=1)[:, 1]
        preds   = outputs.argmax(dim=1)

        all_preds.extend(preds.cpu().numpy())
        all_labels.extend(labels.numpy())
        all_probs.extend(probs.cpu().numpy())

all_preds  = np.array(all_preds)
all_labels = np.array(all_labels)
all_probs  = np.array(all_probs)

print("Stage 1 — Test Set Results")
print("=" * 45)
print(classification_report(all_labels, all_preds, target_names=['Benign', 'Malignant']))
print(f"AUC-ROC: {roc_auc_score(all_labels, all_probs):.4f}")
```

```python
# ── Stage 2 — Filter Malignant Cases Only ─────────────────────────────────────
# Stage 2 only sees mel, bcc, akiec
malignant_df = df[df['dx'].isin(['mel', 'bcc', 'akiec'])].copy()

# Encode the 3 classes as numeric labels
class_map = {'mel': 0, 'bcc': 1, 'akiec': 2}
malignant_df['stage2_label'] = malignant_df['dx'].map(class_map)

print(f"Total malignant samples: {len(malignant_df)}")
print(f"\nClass distribution:")
print(malignant_df['dx'].value_counts())
print(f"\nLabel mapping: {class_map}")

# Split into train/val/test using same random_state for reproducibility
train_s2, temp_s2 = train_test_split(
    malignant_df,
    test_size=0.30,
    random_state=42,
    stratify=malignant_df['stage2_label']
)
val_s2, test_s2 = train_test_split(
    temp_s2,
    test_size=0.50,
    random_state=42,
    stratify=temp_s2['stage2_label']
)

print(f"\nTrain size:      {len(train_s2)}")
print(f"Validation size: {len(val_s2)}")
print(f"Test size:       {len(test_s2)}")
```

```python
# ── Stage 2 — Datasets & DataLoaders ──────────────────────────────────────────
train_dataset_s2 = SkinLesionDataset(train_s2, transform=train_transform,    label_col='stage2_label')
val_dataset_s2   = SkinLesionDataset(val_s2,   transform=val_test_transform, label_col='stage2_label')
test_dataset_s2  = SkinLesionDataset(test_s2,  transform=val_test_transform, label_col='stage2_label')

# Class weights for 3-class imbalance
class_weights_s2 = compute_class_weight(
    class_weight='balanced',
    classes=np.array([0, 1, 2]),
    y=train_s2['stage2_label'].values
)
class_weights_s2 = torch.FloatTensor(class_weights_s2)
print(f"Class weights — mel: {class_weights_s2[0]:.3f}, bcc: {class_weights_s2[1]:.3f}, akiec: {class_weights_s2[2]:.3f}")

# WeightedRandomSampler
sample_weights_s2 = train_s2['stage2_label'].apply(
    lambda x: class_weights_s2[x].item()
).values
sampler_s2 = WeightedRandomSampler(
    weights=sample_weights_s2,
    num_samples=len(sample_weights_s2),
    replacement=True
)

train_loader_s2 = DataLoader(train_dataset_s2, batch_size=32, sampler=sampler_s2)
val_loader_s2   = DataLoader(val_dataset_s2,   batch_size=32, shuffle=False)
test_loader_s2  = DataLoader(test_dataset_s2,  batch_size=32, shuffle=False)

print(f"\n✓ Train loader:      {len(train_loader_s2)} batches")
print(f"✓ Validation loader: {len(val_loader_s2)} batches")
print(f"✓ Test loader:       {len(test_loader_s2)} batches")
```

```python
# ── MobileNetV2 Stage 2 — Multi-class Classifier ──────────────────────────────
model_stage2 = models.mobilenet_v2(weights=models.MobileNet_V2_Weights.DEFAULT)

# Freeze all layers
for param in model_stage2.parameters():
    param.requires_grad = False

# Replace head for 3-class output
model_stage2.classifier = nn.Sequential(
    nn.Dropout(p=0.2),
    nn.Linear(1280, 3)
)

model_stage2 = model_stage2.to(device)

# Loss and optimiser
criterion_s2      = nn.CrossEntropyLoss(weight=class_weights_s2.to(device))
optimizer_s2_warmup = torch.optim.Adam(model_stage2.classifier.parameters(), lr=1e-4)

trainable = sum(p.numel() for p in model_stage2.parameters() if p.requires_grad)
print(f"✓ Stage 2 model ready")
print(f"  Trainable parameters: {trainable:,}")
print(f"✓ Loss: CrossEntropyLoss with class weights")
print(f"✓ Optimiser: Adam (head only, lr=1e-4)")
```

```python
# ── Stage 2 Training Loop ─────────────────────────────────────────────────────
from sklearn.metrics import f1_score

def train_one_epoch_s2(model, loader, optimizer, criterion, device):
    model.train()
    running_loss = 0.0
    all_preds, all_labels = [], []

    for images, labels in loader:
        images, labels = images.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = model(images)
        loss    = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item()
        preds = outputs.argmax(dim=1).cpu().numpy()
        all_preds.extend(preds)
        all_labels.extend(labels.cpu().numpy())

    epoch_loss = running_loss / len(loader)
    epoch_f1   = f1_score(all_labels, all_preds, average='weighted')
    return epoch_loss, epoch_f1


def evaluate_s2(model, loader, criterion, device):
    model.eval()
    running_loss = 0.0
    all_preds, all_labels = [], []

    with torch.no_grad():
        for images, labels in loader:
            images, labels = images.to(device), labels.to(device)
            outputs = model(images)
            loss    = criterion(outputs, labels)

            running_loss += loss.item()
            preds = outputs.argmax(dim=1).cpu().numpy()
            all_preds.extend(preds)
            all_labels.extend(labels.cpu().numpy())

    epoch_loss = running_loss / len(loader)
    epoch_f1   = f1_score(all_labels, all_preds, average='weighted')
    return epoch_loss, epoch_f1


# ── Warm-up Training ──────────────────────────────────────────────────────────
WARMUP_EPOCHS_S2 = 5
best_f1_s2       = 0.0
history_s2       = {'train_loss': [], 'val_loss': [], 'train_f1': [], 'val_f1': []}

print("Starting Stage 2 warm-up training (head only)...")
print(f"{'Epoch':<8} {'Train Loss':<12} {'Train F1':<12} {'Val Loss':<12} {'Val F1':<12}")
print("-" * 55)

for epoch in range(1, WARMUP_EPOCHS_S2 + 1):
    train_loss, train_f1 = train_one_epoch_s2(model_stage2, train_loader_s2, optimizer_s2_warmup, criterion_s2, device)
    val_loss,   val_f1   = evaluate_s2(model_stage2, val_loader_s2, criterion_s2, device)

    history_s2['train_loss'].append(train_loss)
    history_s2['val_loss'].append(val_loss)
    history_s2['train_f1'].append(train_f1)
    history_s2['val_f1'].append(val_f1)

    if val_f1 > best_f1_s2:
        best_f1_s2 = val_f1
        torch.save(model_stage2.state_dict(), 'best_model_stage2.pt')
        saved = "✓ saved"
    else:
        saved = ""

    print(f"{epoch:<8} {train_loss:<12.4f} {train_f1:<12.4f} {val_loss:<12.4f} {val_f1:<12.4f} {saved}")

print(f"\nWarm-up complete. Best validation F1: {best_f1_s2:.4f}")
```

```python
# ── Stage 2 Fine-tuning ───────────────────────────────────────────────────────
for param in model_stage2.features[15:].parameters():
    param.requires_grad = True

trainable = sum(p.numel() for p in model_stage2.parameters() if p.requires_grad)
print(f"✓ Unfrozen blocks 15-18")
print(f"  Trainable parameters now: {trainable:,}")

optimizer_s2_finetune = torch.optim.Adam([
    {'params': model_stage2.features[15:].parameters(), 'lr': 1e-5},
    {'params': model_stage2.classifier.parameters(),    'lr': 1e-4}
])

FINETUNE_EPOCHS_S2 = 10
print(f"\nStarting Stage 2 fine-tune training ({FINETUNE_EPOCHS_S2} epochs)...")
print(f"{'Epoch':<8} {'Train Loss':<12} {'Train F1':<12} {'Val Loss':<12} {'Val F1':<12}")
print("-" * 55)

for epoch in range(1, FINETUNE_EPOCHS_S2 + 1):
    train_loss, train_f1 = train_one_epoch_s2(model_stage2, train_loader_s2, optimizer_s2_finetune, criterion_s2, device)
    val_loss,   val_f1   = evaluate_s2(model_stage2, val_loader_s2, criterion_s2, device)

    history_s2['train_loss'].append(train_loss)
    history_s2['val_loss'].append(val_loss)
    history_s2['train_f1'].append(train_f1)
    history_s2['val_f1'].append(val_f1)

    if val_f1 > best_f1_s2:
        best_f1_s2 = val_f1
        torch.save(model_stage2.state_dict(), 'best_model_stage2.pt')
        saved = "✓ saved"
    else:
        saved = ""

    print(f"{epoch:<8} {train_loss:<12.4f} {train_f1:<12.4f} {val_loss:<12.4f} {val_f1:<12.4f} {saved}")

print(f"\nFine-tuning complete. Best validation F1: {best_f1_s2:.4f}")
```

```python
# ── Stage 2 — Test Set Evaluation ─────────────────────────────────────────────
model_stage2.load_state_dict(torch.load("best_model_stage2.pt", map_location=device))
model_stage2.eval()

all_preds_s2, all_labels_s2, all_probs_s2 = [], [], []

with torch.no_grad():
    for images, labels in test_loader_s2:
        images  = images.to(device)
        outputs = model_stage2(images)
        probs   = torch.softmax(outputs, dim=1)  # shape (batch, 3)
        preds   = outputs.argmax(dim=1)

        all_preds_s2.extend(preds.cpu().numpy())
        all_labels_s2.extend(labels.numpy())
        all_probs_s2.extend(probs.cpu().numpy())

all_preds_s2  = np.array(all_preds_s2)
all_labels_s2 = np.array(all_labels_s2)
all_probs_s2  = np.array(all_probs_s2)   # shape (n_samples, 3)

print("Stage 2 — Test Set Results")
print("=" * 50)
print(classification_report(
    all_labels_s2, all_preds_s2,
    target_names=["Melanoma", "BCC", "AKIEC"]
))
```
